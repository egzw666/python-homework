```python
#1 score_level 成绩等级，超出0‑100抛异常
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须0‑100")
    if score >=90:
        return "优秀"
    elif score >=80:
        return "良好"
    elif score >=60:
        return "及格"
    else:
        return "不及格"

print(score_level(95))
print(score_level(82))
print(score_level(70))
print(score_level(50))
# print(score_level(105)) #打开测试异常

#2 for、while求1‑100和
sum_for =0
for i in range(1,101):
    sum_for +=i
print("for求和1‑100：",sum_for)

sum_while=0
i=1
while i<=100:
    sum_while +=i
    i +=1
print("while求和1‑100：",sum_while)

#3 迭代版阶乘、递归版阶乘；0! =1
def fact_iter(n):
    res=1
    for i in range(2,n+1):
        res *=i
    return res

def fact_recur(n):
    if n<=1:
        return 1
    return n * fact_recur(n-1)

for n in [0,1,5,10]:
    print(f"n={n},迭代:{fact_iter(n)},递归:{fact_recur(n)}")

#4 summary 返回最小、最大、平均值元组
def summary(values):
    return min(values), max(values), sum(values)/len(values)

t = summary([10,20,30])
mi,ma,avg = t #元组解包
print(mi,ma,avg)

#5 sum_all(*args)可变参数
def sum_all(*args):
    return sum(args)

print("无参",sum_all())
print("单参",sum_all(10))
print("多参",sum_all(1,2,3,4))
```

```python
# 欧几里得算法求最大公约数 gcd
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

print("gcd(24,18) =", gcd(24,18))
print("gcd(60,24) =", gcd(60,24))

# 判断是不是素数
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

for x in [2, 7, 10, 13, 21]:
    print(f"{x} 是否素数：{is_prime(x)}")
```

```python
class Student:
    def __init__(self, name, sid, score):
        self.name = name
        self.sid = sid
        self.score = score
    
    def show(self):
        print(f"姓名:{self.name}, 学号:{self.sid}, 分数:{self.score}")
    
    def get_level(self):
        if self.score >=90:
            return "优秀"
        elif self.score >=80:
            return "良好"
        elif self.score >=60:
            return "及格"
        else:
            return "不及格"

s1 = Student("陈厚浩", 20244080122, 86)
s1.show()
print("等级：", s1.get_level())
```
