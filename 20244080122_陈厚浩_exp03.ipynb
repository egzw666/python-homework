```python
import re

text1 = "cat on mat; later: cats and dogs"
print("匹配数字 \\d+：", re.findall(r"\d+", text1))
print("匹配空白 \\s+：", re.findall(r"\s+", text1))
print("匹配单词边界cat \\bcat\\b：", re.findall(r"\bcat\b", text1))
```

```python
text2 = "cat dog bird CAT"
print("cat|dog 匹配：", re.findall(r"cat|dog", text2))

print("\nre.findall(r'a|aa|aaa', 'aaaa')：", re.findall(r"a|aa|aaa", "aaaa"))
print("re.findall(r'aaa|aa|a', 'aaaa')：", re.findall(r"aaa|aa|a", "aaaa"))
```

```python
text3 = "ct cat caat caaat"
print("ca*t：", re.findall(r"ca*t", text3))
print("ca+t：", re.findall(r"ca+t", text3))
print("ca?t：", re.findall(r"ca?t", text3))
print("ca{2}t：", re.findall(r"ca{2}t", text3))
print("ca{1,2}t：", re.findall(r"ca{1,2}t", text3))
```

量词含义：
- ca*t：a 出现 0 次或多次，匹配 ct、cat、caat、caaat
- ca+t：a 出现 1 次或多次，匹配 cat、caat、caaat，不匹配 ct
- ca?t：a 出现 0 次或 1 次，匹配 ct、cat
- ca{2}t：a 恰好出现 2 次，匹配 caat
- ca{1,2}t：a 出现 1~2 次，匹配 cat、caat


```python
text_email = "联系 : alice@example.com ; 备选 : bob@test.org"
pattern = r"(\w+)@(\w+\.\w+)"
result = re.findall(pattern, text_email)
print("邮箱用户名和域名：", result)
for user, domain in result:
    print(f"用户名：{user}，域名：{domain}")
```

```python
text_repeat = "cat cat, dog cat, bird bird"
pattern = r"\b(\w+)\s+\1\b"
print("连续重复单词：", re.findall(pattern, text_repeat))
```

```python
date_text = "2024-05-01\n2024/05/02\n2024.05.03"
print("转换前：")
print(date_text)

new_date = re.sub(r"(\d{4})[-/.](\d{2})[-/.](\d{2})", r"\2-\3-\1", date_text)
print("\n转换后：")
print(new_date)
```

```python
phone_text = "联系我：123-456-7890 或 987.654.3210"
print("转换前：", phone_text)

new_phone = re.sub(r"(\d{3})[-.](\d{3})[-.](\d{4})", r"(\1) \2-\3", phone_text)
print("转换后：", new_phone)
```

```python
from bs4 import BeautifulSoup
print("bs4 可用")
```

```python
html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式
<p class="course">第七讲：结构化数据
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")

# 页面标题文本
print("页面标题：", soup.title.text)
# h1 标签名和文本
h1_tag = soup.find("h1")
print("h1 标签名：", h1_tag.name)
print("h1 文本：", h1_tag.text)
```

```python
# 所有 class="course" 的段落
course_ps = soup.find_all("p", class_="course")
print("所有 course 段落：")
for p in course_ps:
    print(" -", p.text)

# 所有链接
links = soup.find_all("a")
print("\n所有链接：")
for a in links:
    print(f"文字：{a.text}，href：{a['href']}")
```

```python
print("整个文档可见文本：")
print(soup.get_text())

print("\n单个段落文本：")
print(course_ps[0].get_text())

print("\n单个链接文本：")
print(links[0].get_text())
```
