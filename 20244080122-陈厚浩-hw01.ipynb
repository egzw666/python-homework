```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m


# 测试三组数字
print(gcd(2018, 2019))   # 互质，结果1
print(gcd(1200, 300))    # 300
print(gcd(5040, 60))     # 60

# 测试负数
print(gcd(-24, -18))
```

### 问题：gcd输入负数会发生什么？是否合理？
当传入负数时，程序会继续执行取余运算，返回负的最大公约数。
数学上最大公约数的定义仅针对非负整数，结果应当恒为正数。所以虽然代码能算出数值，但结果不符合数学定义，直接传入负数是不合理的。


```python
# 1. euler_limit
def euler_limit(n):
    return (1 + 1 / n) ** n


# 2. euler_infinite_sum
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1
    for k in range(n):
        if k > 0:
            fact *= k
        total += 1 / fact
    return total

# 测试样例
print(euler_infinite_sum(4))


# 3. euler_approx
def euler_approx(eps):
    import math
    total = 0.0
    fact = 1
    k = 0
    while True:
        term = 1.0 / fact
        total += term
        if term < eps:
            break
        k += 1
        fact *= k
    return total

print("真实e:", math.e)
print("epsilon=1e-5近似e:", euler_approx(1e-5))


#4. print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")

print_euler_sum_table(6)
```

```python
#1 is_power_of_2
def is_power_of_2(n):
    # n正整数；判断是否是2的幂
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

print(is_power_of_2(8))   # True
print(is_power_of_2(10))  # False
print(is_power_of_2(1))   # True


#2 is_power(b,n) 判断n是否是b的幂
def is_power(b, n):
    if n == 1:
        return True  # b^0 =1
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

print(is_power(3, 81)) #True
print(is_power(5, 30)) #False
```
